# **DATA AUGMENTATION**

```python
# Converting the "time" column in the accelerometer DataFrame to datetime format
# The 'unit="ns"' parameter specifies that the time values are in nanoseconds
df_Acc.index = pd.to_datetime(df_Acc["time"], unit="ns")
df_Acc.head(1)
```

```python
# Converting the "time" column in the orientation DataFrame to datetime format
# The 'unit="ns"' parameter specifies that the time values are in nanoseconds
df_Or.index = pd.to_datetime(df_Or["time"], unit="ns")
df_Or.head(1)
```

```python
# Removing the "time" and "seconds_elapsed" columns from the accelerometer DataFrame
del df_Acc["time"]
del df_Acc["seconds_elapsed"]

# Removing the "time" and "seconds_elapsed" columns from the orientation DataFrame
del df_Or["time"]
del df_Or["seconds_elapsed"]
```

```python
# Concatenating selected columns from accelerometer (first three columns) with orientation DataFrame
# The resulting DataFrame combines relevant features for further analysis
df = pd.concat([df_Acc.iloc[:, :3], df_Or], axis=1)
df.head()
```

```python
df.columns
```

```python
df.columns = ['acc_z', 'acc_y', 'acc_x', 'or_qz', 'or_qy', 'or_qx', 'or_qw', 'or_roll', 'or_pitch', 'or_yaw', 'Class']
```

```python
df
```

```python
# Define a dictionary 'sampling' specifying aggregation functions for each column during resampling
# For accelerometer data, use the mean value
# For orientation data, use the mean value
# For the "Class" column, use the last value
sampling = {
    "acc_z": "mean",
    "acc_y": "mean",
    "acc_x": "mean",
    "or_qz": "mean",
    "or_qy": "mean",
    "or_qx": "mean",
    "or_qw": "mean",
    "or_roll": "mean",
    "or_pitch": "mean",
    "or_yaw": "mean",
    "Class": "last",
}
```

```python
# Resampling the DataFrame at a 10 milliseconds interval using the specified aggregation functions
# The 'rule="10ms"' parameter sets the resampling rule to a 10 milliseconds interval
# The 'apply(sampling)' part applies the aggregation functions defined in the 'sampling' dictionary
# The 'dropna()' method removes any resulting NaN values
df_resampled = df.resample(rule="10ms").apply(sampling).dropna()
```

# **FEATURE EXTRACTION**

```python
# Define the window size for feature extraction
window_size = 100

# Initialize lists to store the computed features
means = []   # List to store the mean values for each window
stds = []    # List to store the standard deviation values for each window
mins = []    # List to store the minimum values for each window
maxs = []    # List to store the maximum values for each window

# Divide the data into windows and compute features
for i in range(0, len(df_resampled), window_size):
    window = df_resampled.iloc[i:i+window_size]

    # Check if the window size is reached
    if len(window) == window_size:
        # Compute mean, std, min, and max for each window and append to the respective lists
        means.append(window[['acc_z', 'acc_y', 'acc_x', 'or_qz', 'or_qy', 'or_qx', 'or_qw', 'or_roll', 'or_pitch', 'or_yaw']].mean().tolist())
        stds.append(window[['acc_z', 'acc_y', 'acc_x', 'or_qz', 'or_qy', 'or_qx', 'or_qw', 'or_roll', 'or_pitch', 'or_yaw']].std().tolist())
        mins.append(window[['acc_z', 'acc_y', 'acc_x', 'or_qz', 'or_qy', 'or_qx', 'or_qw', 'or_roll', 'or_pitch', 'or_yaw']].min().tolist())
        maxs.append(window[['acc_z', 'acc_y', 'acc_x', 'or_qz', 'or_qy', 'or_qx', 'or_qw', 'or_roll', 'or_pitch', 'or_yaw']].max().tolist())

# Convert the lists into numpy arrays
means = np.array(means)
stds = np.array(stds)
mins = np.array(mins)
maxs = np.array(maxs)

# Create a dataframe for the features
features = pd.DataFrame(np.hstack([means, stds, mins, maxs]), columns=['mean_acc_z', 'mean_acc_y', 'mean_acc_x', 'mean_or_qz', 'mean_or_qy', 'mean_or_qx', 'mean_or_qw', 'mean_or_roll', 'mean_or_pitch', 'mean_or_yaw','std_acc_z', 'std_acc_y', 'std_acc_x', 'std_or_qz', 'std_or_qy', 'std_or_qx', 'std_or_qw', 'std_or_roll', 'std_or_pitch', 'std_or_yaw','min_acc_z', 'min_acc_y', 'min_acc_x', 'min_or_qz', 'min_or_qy', 'min_or_qx', 'min_or_qw', 'min_or_roll', 'min_or_pitch', 'min_or_yaw','max_acc_z', 'max_acc_y', 'max_acc_x', 'max_or_qz', 'max_or_qy', 'max_or_qx', 'max_or_qw', 'max_or_roll', 'max_or_pitch', 'max_or_yaw',])

# Display the shape of the features dataframe
print(features.shape)

# Display the first few rows of the features
features.head(10)
```

# **TRAIN-TEST SPLIT**

```python
# Extracting features (X) and target variable (y) from the resampled DataFrame
X = df_resampled.iloc[:, :10]    # Features: the first 10 columns
y = df_resampled.iloc[:, 10]     # Target variable: "Class" column
```

```python
# Standardizing the features using StandardScaler
scaler = StandardScaler()

# Transforming the features (X) using the fitted StandardScaler
X = scaler.fit_transform(X)
```

```python
# Splitting the dataset into training and testing sets using train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

# Displaying the shapes of the training and testing sets
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
y_train
```

#**SUPPORT VECTOR MACHINE (SVM) WITH DEFAULT HYPERPARAMETERS**

```python
# Instantiate a Support Vector Classification (SVC) classifier with default hyperparameters
svc = SVC()

# Fit the classifier to the training set
svc.fit(X_train, y_train)

# Make predictions on the test set
y_pred = svc.predict(X_test)

# Compute and print accuracy score
print('Model accuracy score with default hyperparameters: {0:0.4f}'.format(accuracy_score(y_test, y_pred)))

# Print confusion matrix
print(confusion_matrix(y_test, y_pred))

# Print classification report
print(classification_report(y_test, y_pred))
```

# **HYPERPARAMETER TUNING FOR SUPPORT VECTOR MACHINE (SVM)**

```python
# Importing GridSearchCV from scikit-learn
from sklearn.model_selection import GridSearchCV

# Defining a parameter grid for hyperparameter tuning
param_grid = {'C': [0.1, 1, 10, 100, 1000],
              'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
              'kernel': ['rbf']}

# Creating a GridSearchCV object with a Support Vector Classification (SVC) estimator
grid = GridSearchCV(SVC(), param_grid, refit=True, verbose=3)

# Fitting the model for grid search using the training set
grid.fit(X_train, y_train)
```

```python
# Print the best parameters after hyperparameter tuning
print(grid.best_params_)

# Print the best estimator (model) after hyperparameter tuning
print(grid.best_estimator_)
```

```python
# Use the best estimator from the grid search to make predictions on the test set
grid_predictions = grid.predict(X_test)

# Print the accuracy score
print(accuracy_score(y_test, grid_predictions))

# Print the confusion matrix
print(confusion_matrix(y_test, grid_predictions))

# Print the classification report to evaluate the model's performance
print(classification_report(y_test, grid_predictions))
```

# **Maximum Accuracy achieved in Support Vector Machine after Hyperparameter tuning is 0.9531**
## **The Parameters which contributed to this accuracy score are**


*   **n_estimators = 25**
*   **C = 1000**
*   **gamma = 0.1**
*   **kernel = 'rbf'**


#**RANDOM FOREST**
## **TUNING NUMBER OF TREES IN THE FOREST(n_estimators) & MAXIMUM DEPTH OF TREE(max_depth)**

```python
# Define lists of hyperparameters to be tuned
k_values = [1, 3, 10, 25, 50, 100]       # Number of decision trees (estimators)
d_values = [1, 2, 3, 5, 10, 15, 20, 30, 50]  # Maximum depth of each decision tree

# Initialize an empty list to store accuracy scores
accuracy = []

# Nested loop to iterate over different combinations of k and d
for i in k_values:
    for j in d_values:
        print('Number of decision trees (estimators) is =', i)
        print('Maximum depth of each decision tree is =', j)

        # Instantiate the Random Forest Classifier with specified hyperparameters
        rfc = RandomForestClassifier(n_estimators=i, max_depth=j, random_state=0)

        # Fit the model on the training set
        rfc.fit(X_train, y_train)

        # Predict the Test set results
        y_pred = rfc.predict(X_test)

        # Compute accuracy and print evaluation metrics
        r_accuracy = accuracy_score(y_test, y_pred)
        r_confusion_matrix = confusion_matrix(y_test, y_pred)
        r_classification_report = classification_report(y_test, y_pred)

        print(f"Accuracy for {i} decision-trees and max depth {j}: {r_accuracy:.4f}")
        print(f"Confusion Matrix for {i} decision trees and max depth {j}:\n {r_confusion_matrix}")
        print(f"Classification Report for {i} decision trees and max depth {j}:\n {r_classification_report}")

        # Append accuracy to the list
        accuracy.append(r_accuracy)
```
