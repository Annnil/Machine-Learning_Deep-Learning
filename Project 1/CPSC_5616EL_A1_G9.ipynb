```python
# Import the KNeighborsClassifier from scikit-learn
from sklearn.neighbors import KNeighborsClassifier

# Import the StandardScaler from scikit-learn for feature scaling
from sklearn.preprocessing import StandardScaler

# Import classification_report and confusion_matrix from scikit-learn for model evaluation
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Import train_test_split from scikit-learn for splitting the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Set a random seed for reproducibility (seed value can be changed)
seed = 42
```

```python
# Select the features (independent variables) from the DataFrame 'df'
X = df.iloc[:,1:23]

# Select the target variable (dependent variable) from the DataFrame 'df'
y = df.iloc[:,0]
```

```python
X.head()
```

```python
y.head()
```

```python
# Split the data into training and testing sets for features (X) and target (y)
# with a test size of 20% (0.2), and using the specified random seed for reproducibility

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=seed)
```

```python
# Import the math module to perform mathematical operations
import math

# Calculate the value of K which is equal to square root of the number of samples in the training data
k_value = math.sqrt(len(X_train))
k_value
```

```python
# Initialize a KNeighborsClassifier with 81 neighbors
knn = KNeighborsClassifier(n_neighbors=81)

# Fit the KNN classifier to the training data
knn.fit(X_train,y_train)
```

```python
# Use the trained KNN classifier to make predictions on the test data

y_pred = knn.predict(X_test)
```

```python
# Print classification report and confusion matrix to evaluate model performance

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
# Import the GridSearchCV class from scikit-learn for hyperparameter tuning

from sklearn.model_selection import GridSearchCV
```

```python
# Define a dictionary 'knn_params' with hyperparameter values and their ranges

knn_params = {
    "n_neighbors": range(71, 90, 2),        # Number of neighbors (odd numbers from 71 to 89)
    "weights": ["uniform", "distance"],     # Weighting schemes for neighbors
    "metric": ["euclidean", "minkowski"],   # Distance metric choices
    "leaf_size": range(1, 50, 10)           # Leaf size for tree-based queries
}
```

```python
# Initialize a KNeighborsClassifier with default hyperparameters

knn1 = KNeighborsClassifier()
```

```python
# Create a GridSearchCV object for hyperparameter tuning
# - 'knn1': The K-nearest neighbors classifier with default hyperparameters
# - 'param_grid': The hyperparameter grid defined in 'knn_params'
# - 'scoring': The evaluation metric to optimize for (accuracy in this case)
# - 'cv': The number of cross-validation folds (5-fold cross-validation)

grid_search = GridSearchCV(knn1, param_grid=knn_params, scoring = "accuracy", cv=5)
```

```python
# Perform the grid search to find the best hyperparameters for the KNN model

grid_results = grid_search.fit(X_train, y_train)
```

```python
# Set the hyperparameters of the KNN model to the best parameters found during grid search
final_model = knn1.set_params(**grid_results.best_params_)

# Fit the final KNN model with the best hyperparameters to the training data
final_model.fit(X_train, y_train)

# Use the final KNN model to make predictions on the test data
y_pred = final_model.predict(X_test)
```

```python
# Print a classification report and confusion matrix to evaluate model performance on the test data

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))

# Print the best hyperparameters found during the grid search

print(grid_results.best_params_)
```
